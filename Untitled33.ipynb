```python
import torch
import random
import numpy as np
import pandas as pd

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM
)

import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Using Device:", device)
print("PyTorch Version:", torch.__version__)
```

```python
model_id = "openai-community/gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_id)

tokenizer.pad_token = tokenizer.eos_token

# Load original pretrained model
base_model = AutoModelForCausalLM.from_pretrained(
    model_id
).to(device)

base_model.eval()

print("GPT-2 Model Loaded Successfully!")
```

```python

def continue_text(
    prompt,
    model=None,
    sample=False,
    temperature=0.8,
    seed=42
):

    if model is None:
        model = base_model

    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(device)

    options = {
        "max_new_tokens": 45,
        "pad_token_id": tokenizer.eos_token_id
    }

    if sample:
        options.update({
            "do_sample": True,
            "temperature": temperature,
            "top_p": 0.9
        })
    else:
        options["do_sample"] = False

    model.eval()

    with torch.no_grad():

        output = model.generate(
            **inputs,
            **options
        )

    return tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )
```

```python

prompts = [
    "A story about a robot begins like this:",
    "In a school library, a tiny robot found a lost book. It"
]

for prompt in prompts:

    print("\nPROMPT:", prompt)

    print("\nGREEDY OUTPUT:")
    print(continue_text(prompt))

    print("\nSAMPLING OUTPUTS:")

    for temp in [0.6, 1.0]:

        for seed in [42, 43]:

            result = continue_text(
                prompt,
                sample=True,
                temperature=temp,
                seed=seed
            )

            print(
                f"\nTemperature={temp}, Seed={seed}"
            )

            print(result)
```

```python
examples = [
    ("graphs",
     "Sketch a graph, then mark a path through three vertices."),

    ("Python loops",
     "Trace a loop by writing the counter after each step."),

    ("fractions",
     "Draw equal parts and label the fraction you shaded."),

    ("sets",
     "Draw a Venn diagram and label the intersection."),

    ("sorting",
     "Sort five values by hand and record each swap."),

    ("probability",
     "List the outcomes of two coins before counting.")
]

print("Training Examples:")

for topic, tip in examples:
    print(topic, ":", tip)
```

```python
tuned = AutoModelForCausalLM.from_pretrained(
    model_id
).to(device)

# Freeze all model parameters
for param in tuned.parameters():
    param.requires_grad = False

# Unfreeze only the last Transformer block
for param in tuned.transformer.h[-1].parameters():
    param.requires_grad = True

# Optimizer
optimizer = torch.optim.AdamW(
    tuned.transformer.h[-1].parameters(),
    lr=2e-5
)

trainable = sum(
    p.numel() for p in tuned.parameters()
    if p.requires_grad
)

print("Trainable Parameters:", trainable)
print("Only the last Transformer block is trainable.")
```

```python
epochs = 5

loss_history = []

for epoch in range(epochs):

    tuned.train()

    epoch_losses = []

    for topic, answer in examples:

        # Prepare prompt
        prompt = f"Topic: {topic}\nStudy tip:"

        prefix_ids = tokenizer.encode(
            prompt,
            add_special_tokens=False
        )

        target_ids = tokenizer.encode(
            " " + answer,
            add_special_tokens=False
        )

        # Add EOS token
        target_ids = target_ids + [
            tokenizer.eos_token_id
        ]

        # Combine prompt and target
        input_ids = torch.tensor(
            [prefix_ids + target_ids],
            device=device
        )

        # Ignore prompt tokens in loss
        labels = torch.tensor(
            [[-100] * len(prefix_ids) + target_ids],
            device=device
        )

        optimizer.zero_grad()

        outputs = tuned(
            input_ids=input_ids,
            labels=labels
        )

        loss = outputs.loss

        loss.backward()

        optimizer.step()

        epoch_losses.append(loss.item())

    avg_loss = np.mean(epoch_losses)

    loss_history.append(avg_loss)

    print(
        f"Epoch {epoch + 1}/{epochs} "
        f"Loss: {avg_loss:.4f}"
    )

# Save fine-tuned model
tuned.save_pretrained("study_tip_gpt2")

tokenizer.save_pretrained("study_tip_gpt2")

print("Fine-tuning Completed!")
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, epochs + 1),
    loss_history,
    marker="o",
    label="Training Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")

plt.title("GPT-2 Fine-Tuning Loss")

plt.xticks(range(1, epochs + 1))

plt.grid()
plt.legend()

plt.savefig(
    "gpt2_finetuning_loss.png",
    dpi=300
)

plt.show()
```

```python
held_out = [
    "recursion",
    "matrices",
    "binary search"
]

base_outputs = {}
tuned_outputs = {}

for topic in held_out:

    prompt = f"Topic: {topic}\nStudy tip:"

    print("\n================================")

    print("HELD-OUT TOPIC:", topic)

    print("\nBASE GPT-2 OUTPUT:")

    base_text = continue_text(
        prompt,
        model=base_model,
        sample=False
    )

    print(base_text)

    print("\nFINE-TUNED GPT-2 OUTPUT:")

    tuned_text = continue_text(
        prompt,
        model=tuned,
        sample=False
    )

    print(tuned_text)

    base_outputs[topic] = base_text
    tuned_outputs[topic] = tuned_text
```

```python
evaluation = pd.DataFrame({
    "Held-out Topic": held_out,
    "Base Score out of 4": ["Fill after evaluation"] * 3,
    "Tuned Score out of 4": ["Fill after evaluation"] * 3,
    "Example Difference": [
        "Compare the generated advice",
        "Compare the generated advice",
        "Compare the generated advice"
    ]
})

display(evaluation)
```

```python

comparison = pd.DataFrame({
    "Topic": held_out,
    "Base GPT-2": [
        base_outputs[t] for t in held_out
    ],
    "Fine-Tuned GPT-2": [
        tuned_outputs[t] for t in held_out
    ]
})

pd.set_option("display.max_colwidth", 100)

display(comparison)

comparison.to_csv(
    "gpt2_comparison.csv",
    index=False
)
```
